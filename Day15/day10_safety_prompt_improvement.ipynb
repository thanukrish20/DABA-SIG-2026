```python
# Day 10 - Safety Prompt Improvement

safety_rules = [
    "Do not judge or blame the user.",
    "Do not make assumptions about the user's feelings.",
    "Do not provide a medical diagnosis.",
    "Use supportive and respectful language.",
    "Do not ignore safety-related concerns.",
    "Encourage reaching out to a trusted person when needed.",
    "Keep responses relevant to the journal entry.",
    "If a safety concern is detected, prioritize supportive guidance."
]

print("DAY 10 - IMPROVED SAFETY RULES\n")

for number, rule in enumerate(safety_rules, start=1):
    print(f"{number}. {rule}")
```

```python
safe_prompts = {
    "Positive": (
        "Respond warmly. Acknowledge the user's experience "
        "without making unsupported assumptions."
    ),
    "Neutral": (
        "Respond calmly. Ask a simple, relevant reflection question "
        "without forcing a positive interpretation."
    ),
    "Negative": (
        "Acknowledge that the day may feel difficult. "
        "Be respectful and supportive. Do not judge or diagnose."
    ),
    "Safety": (
        "Prioritize the user's safety. Respond calmly and compassionately. "
        "Encourage contacting a trusted person or qualified professional. "
        "If immediate danger exists, advise contacting local emergency services."
    ),
    "Empty": (
        "Ask the user to enter a journal entry before continuing."
    )
}

print("DAY 10 - IMPROVED SAFETY PROMPTS\n")

for category, prompt in safe_prompts.items():
    print(f"{category} Prompt:")
    print(prompt)
    print("-" * 50)
```

```python
def improved_safety_check(journal):
    text = journal.lower()

    safety_keywords = [
        "suicide",
        "kill myself",
        "end my life",
        "self-harm"
    ]

    if any(keyword in text for keyword in safety_keywords):
        return "Safety"

    if not journal.strip():
        return "Empty"

    return "Normal"


def improved_mood_assistant(journal):
    status = improved_safety_check(journal)

    if status == "Empty":
        return safe_prompts["Empty"]

    if status == "Safety":
        return (
            "I'm sorry you're facing something difficult. "
            "Please reach out to a trusted person or qualified "
            "support professional. If you are in immediate danger, "
            "contact local emergency services."
        )

    result = analyze_mood(journal)
    mood = result["mood"]

    prompt_category = mood if mood in safe_prompts else "Neutral"
    prompt = safe_prompts[prompt_category]

    return (
        f"Detected mood: {mood}\n"
        f"Safety prompt: {prompt}"
    )


print("Improved safety guardrails are ready!")
```

```python
from textblob import TextBlob

def analyze_mood(journal):
    score = TextBlob(journal).sentiment.polarity

    if score > 0.1:
        mood = "Positive"
    elif score < -0.1:
        mood = "Negative"
    else:
        mood = "Neutral"

    return {
        "sentiment_score": round(score, 2),
        "mood": mood
    }

print("Mood analysis function is ready!")
```
