```python
import pandas as pd

support_data = {
    "category": [
        "Positive",
        "Neutral",
        "Negative",
        "Stress",
        "General"
    ],
    "support_message": [
        "That's great! You can record what made today meaningful.",
        "A normal day is okay too. You may reflect on one thing you noticed today.",
        "Sorry today felt difficult. Consider talking with someone you trust.",
        "Try taking a short break and reaching out to someone supportive if needed.",
        "Your journal is your space to reflect. Share only what you feel comfortable sharing."
    ],
    "suggested_action": [
        "Reflect on positive experiences",
        "Continue journaling if helpful",
        "Reach out to a trusted person",
        "Take a break or seek support",
        "Choose whether to reflect or pause"
    ]
}

support_kb = pd.DataFrame(support_data)

print("SAFE-SUPPORT KNOWLEDGE BASE")
display(support_kb)
```

```python
def check_safety_rules(journal):
    text = journal.lower()

    safety_keywords = [
        "suicide",
        "kill myself",
        "end my life",
        "self-harm"
    ]

    if any(keyword in text for keyword in safety_keywords):
        return {
            "safety_level": "Needs attention",
            "response": (
                "I'm sorry things feel difficult. "
                "Please reach out to a trusted person or "
                "qualified support professional. "
                "If there is immediate danger, contact local "
                "emergency services."
            )
        }

    return {
        "safety_level": "No keyword detected",
        "response": "Continue with the regular support workflow."
    }

print("Safety rules created successfully.")
print(check_safety_rules("Today was stressful and difficult."))
```

```python
def get_support_response(journal):
    safety_result = check_safety_rules(journal)

    if safety_result["safety_level"] == "Needs attention":
        return {
            "category": "Safety Support",
            "safety_level": safety_result["safety_level"],
            "message": safety_result["response"]
        }

    from textblob import TextBlob
    score = TextBlob(journal).sentiment.polarity

    if score > 0.1:
        category = "Positive"
    elif score < -0.1:
        category = "Negative"
    else:
        category = "Neutral"

    selected = support_kb[support_kb["category"] == category].iloc[0]

    return {
        "category": category,
        "safety_level": safety_result["safety_level"],
        "message": selected["support_message"],
        "suggested_action": selected["suggested_action"]
    }

# Test with a sample journal
result = get_support_response(
    "Today was difficult and frustrating."
)

print("Category:", result["category"])
print("Safety status:", result["safety_level"])
print("Support message:", result["message"])
print("Suggested action:", result.get("suggested_action", "Seek trusted support"))
```

```python
test_journals = [
    "I had a wonderful day with my friends.",
    "Today was difficult and frustrating.",
    "I am thinking about suicide and need help."
]

for i, journal in enumerate(test_journals, start=1):
    result = get_support_response(journal)

    print(f"\nTEST {i}")
    print("Journal:", journal)
    print("Category:", result["category"])
    print("Safety status:", result["safety_level"])
    print("Support message:", result["message"])
```

```python
print("===== DAY 4: SAFE-SUPPORT REPORT =====")

print("Knowledge Base Categories:", len(support_kb))
print("Safety Rules: Configured")
print("Support Response Function: Ready")
print("Test Scenarios:", len(test_journals))

for i, journal in enumerate(test_journals, start=1):
    result = get_support_response(journal)
    print(
        f"Test {i}: Category={result['category']} | "
        f"Safety={result['safety_level']}"
    )

print("\nDay 4 implementation completed!")
```
