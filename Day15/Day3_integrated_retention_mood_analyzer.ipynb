```python
import pandas as pd

user_data = {
    "user_id": ["U001", "U002", "U003", "U004", "U005"],
    "login_days": [22, 4, 18, 3, 15],
    "session_minutes": [50, 8, 40, 10, 35],
    "features_used": [8, 2, 7, 1, 6],
    "notifications_opened": [15, 1, 12, 0, 9],
    "journal": [
        "I had a wonderful and happy day.",
        "Today was difficult and frustrating.",
        "I enjoyed learning something new.",
        "I felt unhappy and disappointed today.",
        "My day was normal and peaceful."
    ]
}

df = pd.DataFrame(user_data)

print("DAY 3: APP USAGE + MOOD JOURNAL DATA")
display(df)
```

```python
from textblob import TextBlob

def get_mood(journal):
    score = TextBlob(journal).sentiment.polarity

    if score > 0.1:
        mood = "Positive"
    elif score < -0.1:
        mood = "Negative"
    else:
        mood = "Neutral"

    return round(score, 2), mood

df[["sentiment_score", "mood"]] = df["journal"].apply(
    lambda text: pd.Series(get_mood(text))
)

print("MOOD SENTIMENT ANALYSIS")
display(df[["user_id", "journal", "sentiment_score", "mood"]])
```

```python
def get_user_insight(row):
    if row["login_days"] < 7 and row["mood"] == "Negative":
        return "Low app usage and negative journal sentiment"
    elif row["login_days"] < 7:
        return "Low app usage; review engagement"
    elif row["mood"] == "Negative":
        return "Regular usage with negative journal sentiment"
    else:
        return "Regular usage; continue monitoring"

df["combined_insight"] = df.apply(get_user_insight, axis=1)

print("COMBINED USAGE AND MOOD INSIGHTS")
display(
    df[[
        "user_id",
        "login_days",
        "session_minutes",
        "mood",
        "sentiment_score",
        "combined_insight"
    ]]
)
```

```python
def safety_guardrail(journal):
    text = journal.lower()

    safety_keywords = [
        "suicide",
        "kill myself",
        "end my life",
        "self-harm"
    ]

    if any(keyword in text for keyword in safety_keywords):
        return "Safety flag - human support review recommended"
    else:
        return "No safety keywords detected"

df["safety_status"] = df["journal"].apply(safety_guardrail)

print("USAGE + MOOD + SAFETY REPORT")
display(
    df[[
        "user_id",
        "mood",
        "combined_insight",
        "safety_status"
    ]]
)
```

```python
print("===== DAY 3: INTEGRATED USER REPORT =====")

for _, row in df.iterrows():
    print("\nUser ID:", row["user_id"])
    print("Login Days:", row["login_days"])
    print("Session Minutes:", row["session_minutes"])
    print("Mood:", row["mood"])
    print("Sentiment Score:", row["sentiment_score"])
    print("Combined Insight:", row["combined_insight"])
    print("Safety Status:", row["safety_status"])
    print("-" * 45)

print("\nDay 3 integrated analysis completed!")
```
