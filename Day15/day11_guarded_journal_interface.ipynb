```python
import pandas as pd
from IPython.display import display, Markdown
import ipywidgets as widgets

print("Day 11 - Guarded Journal Interface")
print("Required libraries are ready!")
```

```python
journal_box = widgets.Textarea(
    value="",
    placeholder="Write about your day here...",
    description="Journal:",
    layout=widgets.Layout(width="90%", height="150px")
)

submit_button = widgets.Button(
    description="Analyze Journal",
    button_style="primary",
    icon="check"
)

output_box = widgets.Output()

display(journal_box)
display(submit_button)
display(output_box)
```

```python
def improved_safety_check(journal):
    text = journal.lower().strip()

    if not text:
        return "Empty"

    safety_keywords = [
        "suicide",
        "kill myself",
        "end my life",
        "self-harm"
    ]

    if any(keyword in text for keyword in safety_keywords):
        return "Safety"

    return "Normal"


print("Safety check function is ready!")
```

```python
def on_analyze_click(button):
    with output_box:
        output_box.clear_output()

        journal = journal_box.value.strip()

        if not journal:
            print("Please enter a journal entry.")
            return

        status = improved_safety_check(journal)

        if status == "Safety":
            print("Safety Support")
            print(
                "I'm sorry you're facing something difficult. "
                "Please reach out to a trusted person or qualified "
                "support professional. If you are in immediate danger, "
                "contact local emergency services."
            )
            return

        result = analyze_mood(journal)

        print("Journal Analysis")
        print("Detected text sentiment:", result["mood"])
        print("Sentiment score:", result["sentiment_score"])

        if result["mood"] == "Positive":
            print("Response: What made this experience meaningful to you?")
        elif result["mood"] == "Negative":
            print(
                "Response: It sounds like this may have been a difficult day. "
                "Would talking to someone you trust help?"
            )
        else:
            print(
                "Response: Would you like to reflect on one thing from your day?"
            )


submit_button.on_click(on_analyze_click)

print("Safety guardrails connected to the journal interface!")
```
