```python
import pandas as pd

test_cases = {
    "test_id": [1, 2, 3, 4],
    "journal": [
        "I had a wonderful day with my friends.",
        "Today was an ordinary day.",
        "I feel very upset and need support.",
        ""
    ],
    "expected_condition": [
        "Positive mood response",
        "Neutral mood response",
        "Negative mood response",
        "Empty input handling"
    ]
}

evaluation_df = pd.DataFrame(test_cases)

print("DAY 8 - RESPONSE EVALUATION DATASET")
display(evaluation_df)
```

```python
from textblob import TextBlob

def analyze_mood(journal):
    score = TextBlob(journal).sentiment.polarity

    if score > 0.1:
        mood = "Positive"
    elif score < -0.1:
        mood = "Negative"
    else:
        mood = "Neutral"

    return {
        "sentiment_score": round(score, 2),
        "mood": mood
    }


def guarded_mood_assistant(journal):
    if not journal.strip():
        return "Please enter a journal entry."

    safety_keywords = [
        "suicide", "kill myself",
        "end my life", "self-harm"
    ]

    text = journal.lower()

    if any(word in text for word in safety_keywords):
        return (
            "Safety support response: "
            "Please reach out to a trusted person "
            "or qualified support professional."
        )

    result = analyze_mood(journal)
    mood = result["mood"]

    if mood == "Positive":
        response = "What made this experience meaningful to you?"
    elif mood == "Negative":
        response = (
            "It sounds like this may have been a difficult day. "
            "Would talking to someone you trust help?"
        )
    else:
        response = "Would you like to reflect on one thing from your day?"

    return (
        f"Mood: {mood}\n"
        f"Sentiment score: {result['sentiment_score']}\n"
        f"Supportive response: {response}"
    )

print("Assistant function is ready!")
```

```python
evaluation_results = []

for _, row in evaluation_df.iterrows():
    response = guarded_mood_assistant(row["journal"])

    evaluation_results.append({
        "Test ID": row["test_id"],
        "Expected Condition": row["expected_condition"],
        "Assistant Response": response
    })

results_df = pd.DataFrame(evaluation_results)

print("DAY 8 - RESPONSE EVALUATION")
display(results_df)
```

```python
def evaluate_response(journal, expected):
    response = guarded_mood_assistant(journal)
    text = response.lower()

    if expected == "Empty input handling":
        relevant = response == "Please enter a journal entry."
        safe = relevant

    elif expected == "Positive mood response":
        relevant = "mood: positive" in text
        safe = "meaningful" in text

    elif expected == "Neutral mood response":
        relevant = "mood: neutral" in text
        safe = "reflect" in text

    elif expected == "Negative mood response":
        relevant = "mood: negative" in text
        safe = "trust" in text

    else:
        relevant = False
        safe = False

    return response, relevant, safe


for _, row in evaluation_df.iterrows():
    response, relevant, safe = evaluate_response(
        row["journal"],
        row["expected_condition"]
    )

    print("Test:", row["test_id"])
    print("Condition:", row["expected_condition"])
    print("Relevance:", "Pass" if relevant else "Fail")
    print("Safety compliance:", "Pass" if safe else "Fail")
    print("-" * 40)
```

```python
total_tests = len(evaluation_df)
relevance_pass = 0
safety_pass = 0

for _, row in evaluation_df.iterrows():
    response, relevant, safe = evaluate_response(
        row["journal"],
        row["expected_condition"]
    )

    if relevant:
        relevance_pass += 1

    if safe:
        safety_pass += 1

print("DAY 8 - EVALUATION SUMMARY")
print("Total test cases:", total_tests)
print("Relevance passed:", relevance_pass)
print("Relevance failed:", total_tests - relevance_pass)
print("Safety compliance passed:", safety_pass)
print("Safety compliance failed:", total_tests - safety_pass)
```
