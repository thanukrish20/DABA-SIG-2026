```python
import pandas as pd

test_data = {
    "test_id": [1, 2, 3, 4, 5],
    "condition": [
        "Positive mood",
        "Neutral mood",
        "Negative mood",
        "Safety keyword",
        "Empty input"
    ],
    "journal": [
        "I enjoyed my day and learned something new.",
        "Today was an ordinary day.",
        "Today was difficult and frustrating.",
        "I need help because I am thinking about suicide.",
        ""
    ]
}

test_df = pd.DataFrame(test_data)

print("DAY 7 - JOURNAL TEST DATASET")
display(test_df)
```

```python
from textblob import TextBlob

def analyze_mood(journal):
    score = TextBlob(journal).sentiment.polarity

    if score > 0.1:
        mood = "Positive"
    elif score < -0.1:
        mood = "Negative"
    else:
        mood = "Neutral"

    return {
        "sentiment_score": round(score, 2),
        "mood": mood
    }


def check_journal_safety(journal):
    text = journal.lower()

    safety_keywords = [
        "suicide",
        "kill myself",
        "end my life",
        "self-harm"
    ]

    if any(keyword in text for keyword in safety_keywords):
        return "SAFETY_SUPPORT"

    return "NORMAL"


def guarded_mood_assistant(journal):
    if not journal.strip():
        return "Please enter a journal entry."

    safety_status = check_journal_safety(journal)

    if safety_status == "SAFETY_SUPPORT":
        return (
            "I'm sorry things feel difficult. "
            "Please reach out to a trusted person or qualified "
            "support professional. If you are in immediate danger, "
            "contact local emergency services."
        )

    result = analyze_mood(journal)
    mood = result["mood"]

    if mood == "Positive":
        response = "What made this experience meaningful to you?"
    elif mood == "Negative":
        response = (
            "It sounds like this may have been a difficult day. "
            "Would talking to someone you trust help?"
        )
    else:
        response = "Would you like to reflect on one thing from your day?"

    return (
        f"Mood: {mood}\n"
        f"Sentiment score: {result['sentiment_score']}\n"
        f"Supportive response: {response}"
    )


print("Day 7 assistant functions are ready!")
```

```python
print("DAY 7 - JOURNAL SAFETY TEST RESULTS\n")

for _, row in test_df.iterrows():
    print(f"Test {row['test_id']}: {row['condition']}")

    response = guarded_mood_assistant(row["journal"])
    print("Assistant Response:", response)

    print("-" * 50)
```

```python
passed = 0
failed = 0

for _, row in test_df.iterrows():
    response = guarded_mood_assistant(row["journal"])

    if row["condition"] == "Empty input":
        success = response == "Please enter a journal entry."
    elif row["condition"] == "Safety keyword":
        success = "Please reach out to a trusted person" in response
    else:
        success = "Mood:" in response

    if success:
        passed += 1
    else:
        failed += 1

print("DAY 7 - TEST SUMMARY")
print("Total tests:", len(test_df))
print("Passed:", passed)
print("Failed:", failed)

if failed == 0:
    print("All tests passed!")
else:
    print("Some tests need checking.")
```
