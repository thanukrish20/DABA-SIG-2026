```python
import pandas as pd
from textblob import TextBlob

print("Day 14 - Guarded Journal Assistant Demo")
print("Libraries imported successfully!")
```

```python
usage_data = {
    "user_id": [1, 2, 3, 4, 5],
    "login_days": [6, 2, 5, 1, 7],
    "session_minutes": [45, 10, 35, 8, 60],
    "features_used": [4, 1, 3, 1, 5],
    "notifications_opened": [3, 0, 2, 0, 4]
}

usage_df = pd.DataFrame(usage_data)

usage_df["usage_status"] = usage_df.apply(
    lambda row: "High Usage"
    if row["login_days"] >= 5 and row["session_minutes"] >= 30
    else "Low Usage",
    axis=1
)

display(usage_df)

print("App usage analysis completed!")
```

```python
def analyze_mood(journal):
    score = TextBlob(journal).sentiment.polarity

    if score > 0.1:
        mood = "Positive"
    elif score < -0.1:
        mood = "Negative"
    else:
        mood = "Neutral"

    return mood, round(score, 2)


def check_safety(journal):
    text = journal.lower().strip()

    if not text:
        return "Empty"

    safety_keywords = [
        "suicide",
        "kill myself",
        "end my life",
        "self-harm"
    ]

    if any(word in text for word in safety_keywords):
        return "Safety"

    return "Normal"


def guarded_journal_assistant(journal):
    status = check_safety(journal)

    if status == "Empty":
        return "Please enter a journal entry."

    if status == "Safety":
        return (
            "Safety support is needed. "
            "Please reach out to a trusted person or qualified "
            "support professional. If you are in immediate danger, "
            "contact local emergency services."
        )

    mood, score = analyze_mood(journal)

    if mood == "Positive":
        response = "What made this experience meaningful to you?"
    elif mood == "Negative":
        response = (
            "It sounds like you may be having a difficult day. "
            "Would talking to someone you trust help?"
        )
    else:
        response = "Would you like to reflect on one thing from your day?"

    return {
        "mood": mood,
        "sentiment_score": score,
        "safety_status": status,
        "response": response
    }


print("Guarded journal assistant is ready!")
```

```python
def demonstrate_assistant(user_id, journal):
    print("GUARDED JOURNAL ASSISTANT DEMO")
    print("-" * 35)

    user_data = usage_df[usage_df["user_id"] == user_id]

    if user_data.empty:
        print("User ID not found.")
        return

    print("\nApp Usage Details:")
    display(user_data)

    print("\nJournal Analysis:")
    result = guarded_journal_assistant(journal)

    if isinstance(result, str):
        print(result)
    else:
        print("Detected Mood:", result["mood"])
        print("Sentiment Score:", result["sentiment_score"])
        print("Safety Status:", result["safety_status"])
        print("Supportive Response:", result["response"])


demonstrate_assistant(
    1,
    "I had a good day at college and enjoyed learning new things."
)
```

```python
demo_scenarios = [
    {
        "user_id": 1,
        "journal": "I enjoyed my day and learned something new."
    },
    {
        "user_id": 2,
        "journal": "Today was an ordinary day."
    },
    {
        "user_id": 3,
        "journal": "I feel sad and frustrated today."
    },
    {
        "user_id": 4,
        "journal": ""
    },
    {
        "user_id": 5,
        "journal": "I need help because I am thinking about suicide."
    }
]

for i, scenario in enumerate(demo_scenarios, start=1):
    print(f"\nDEMO SCENARIO {i}")
    print("=" * 35)
    print("User ID:", scenario["user_id"])
    print("Journal:", scenario["journal"] or "(Empty input)")

    result = guarded_journal_assistant(scenario["journal"])
    print("Assistant Result:", result)
```

```python
print("DAY 14 - GUARDED JOURNAL ASSISTANT DEMO REPORT")
print("-" * 50)

print("Total Demo Scenarios:", len(demo_scenarios))

for i, scenario in enumerate(demo_scenarios, start=1):
    result = guarded_journal_assistant(scenario["journal"])

    if isinstance(result, str):
        if result == "Please enter a journal entry.":
            status = "Empty Input Handled"
        else:
            status = "Safety Response Triggered"
    else:
        status = "Journal Analyzed"

    print(f"Scenario {i}: {status}")

print("\nApp usage integration: Available")
print("Journal sentiment analysis: Available")
print("Safety keyword screening: Available")
print("Supportive response generation: Available")
print("\nDemonstration completed!")
```
