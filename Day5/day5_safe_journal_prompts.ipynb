```python
safe_prompts = {
    "Positive": (
        "Acknowledge the user's experience warmly. "
        "Invite them to reflect on what went well, without pressure."
    ),
    "Neutral": (
        "Respond calmly and without judgment. "
        "Invite the user to share more only if they feel comfortable."
    ),
    "Negative": (
        "Acknowledge that the day may have felt difficult. "
        "Offer a gentle reflection question and suggest reaching "
        "out to someone trusted if support would help."
    ),
    "Safety": (
        "Respond with care and without judgment. "
        "Do not diagnose or make assumptions. Encourage the user "
        "to contact a trusted person or qualified support professional. "
        "If immediate danger exists, advise contacting local emergency services."
    )
}

print("SAFE JOURNAL PROMPT TEMPLATES")
for category, prompt in safe_prompts.items():
    print(f"\n{category} Prompt:")
    print(prompt)
```

```python
safety_rules = [
    "Do not diagnose mental health conditions.",
    "Do not judge, shame, or blame the user.",
    "Do not treat sentiment scores as facts about the user.",
    "Do not make assumptions from a single journal entry.",
    "Respect the user's choice not to share more.",
    "Prioritize supportive guidance when safety concerns appear.",
    "Encourage trusted human or professional support when appropriate.",
    "Do not claim keyword screening can guarantee safety."
]

print("JOURNAL SAFETY RULES")
for number, rule in enumerate(safety_rules, start=1):
    print(f"{number}. {rule}")

print("\nTotal safety rules:", len(safety_rules))
```

```python
def select_safe_prompt(journal):
    text = journal.lower()

    safety_keywords = [
        "suicide",
        "kill myself",
        "end my life",
        "self-harm"
    ]

    if any(keyword in text for keyword in safety_keywords):
        category = "Safety"
    else:
        from textblob import TextBlob
        score = TextBlob(journal).sentiment.polarity

        if score > 0.1:
            category = "Positive"
        elif score < -0.1:
            category = "Negative"
        else:
            category = "Neutral"

    return {
        "category": category,
        "prompt": safe_prompts[category]
    }

# Test the prompt selector
sample_journal = "Today was difficult and frustrating."
result = select_safe_prompt(sample_journal)

print("Selected Category:", result["category"])
print("Selected Prompt:", result["prompt"])
```

```python
test_entries = [
    "I had a wonderful day with my friends.",
    "Today was difficult and frustrating.",
    "Today was an ordinary day.",
    "I need help because I am thinking about suicide."
]

print("===== SAFE JOURNAL PROMPT TESTING =====")

for i, journal in enumerate(test_entries, start=1):
    result = select_safe_prompt(journal)

    print(f"\nTest {i}")
    print("Journal:", journal)
    print("Selected Category:", result["category"])
    print("Prompt:", result["prompt"])
    print("-" * 40)
```

```python
def journal_interaction():
    journal = input("Write your journal entry: ").strip()

    if not journal:
        print("Please enter a journal entry.")
        return

    result = select_safe_prompt(journal)

    print("\n===== JOURNAL ANALYSIS =====")
    print("Selected Category:", result["category"])

    print("\nSafety Guidelines:")
    for rule in safety_rules:
        print("-", rule)

    print("\nSelected Response Prompt:")
    print(result["prompt"])

    print("\nNote: This is a basic text analysis tool,")
    print("not a mental-health diagnosis or a substitute for professional support.")

journal_interaction()
```
