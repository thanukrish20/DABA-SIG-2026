```python
from textblob import TextBlob

def analyze_mood(journal):
    score = TextBlob(journal).sentiment.polarity

    if score > 0.1:
        mood = "Positive"
    elif score < -0.1:
        mood = "Negative"
    else:
        mood = "Neutral"

    return {
        "sentiment_score": round(score, 2),
        "mood": mood
    }

# Test the function
sample_journal = "Today was a good day. I enjoyed learning Python."
result = analyze_mood(sample_journal)

print("Journal:", sample_journal)
print("Sentiment Score:", result["sentiment_score"])
print("Detected Mood:", result["mood"])
```

```python
def check_journal_safety(journal):
    text = journal.lower()

    safety_keywords = [
        "suicide",
        "kill myself",
        "end my life",
        "self-harm"
    ]

    if any(keyword in text for keyword in safety_keywords):
        return "SAFETY_SUPPORT"

    return "NORMAL"

# Test the safety check
sample_journal = "Today was a good day at college."

print("Safety Status:", check_journal_safety(sample_journal))
```

```python
def guarded_mood_assistant(journal):
    if not journal.strip():
        return "Please enter a journal entry."

    safety_status = check_journal_safety(journal)

    if safety_status == "SAFETY_SUPPORT":
        return (
            "I'm sorry things feel difficult. "
            "Please reach out to a trusted person or qualified "
            "support professional. If you are in immediate danger, "
            "contact local emergency services."
        )

    mood_result = analyze_mood(journal)
    mood = mood_result["mood"]

    if mood == "Positive":
        response = (
            "Thank you for sharing. What made this experience "
            "meaningful to you?"
        )
    elif mood == "Negative":
        response = (
            "It sounds like this entry may describe a difficult day. "
            "Would reflecting on it or talking to someone you trust help?"
        )
    else:
        response = (
            "Thank you for sharing. Would you like to reflect "
            "on one thing from your day?"
        )

    return (
        f"Detected text sentiment: {mood}\n"
        f"Sentiment score: {mood_result['sentiment_score']}\n"
        f"Supportive response: {response}"
    )

# Test the assistant
print(guarded_mood_assistant(
    "Today was difficult, but I finished my college work."
))
```

```python
test_entries = [
    "I had a wonderful day with my friends.",
    "Today was difficult and frustrating.",
    "Today was an ordinary day.",
    "I need help because I am thinking about suicide."
]

for i, journal in enumerate(test_entries, start=1):
    print(f"\n===== TEST {i} =====")
    print("Journal:", journal)
    print("Assistant Response:")
    print(guarded_mood_assistant(journal))
    print("-" * 40)
```

```python
print("===== GUARDED MOOD JOURNAL ASSISTANT =====")

journal = input("How was your day? Write your journal entry: ")

response = guarded_mood_assistant(journal)

print("\n===== ASSISTANT RESPONSE =====")
print(response)
```
